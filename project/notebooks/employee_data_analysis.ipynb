## Dataset
This project uses the trainer-provided Day 3 employees.csv file. The local dataset is stored in project/data/ and excluded from Git.

## Load the data

```python
from pathlib import Path
import pandas as pd

candidate_paths = [
    Path.cwd() / 'project' / 'data' / 'employees.csv',
    Path.cwd() / 'data' / 'employees.csv',
    Path.cwd().parent / 'data' / 'employees.csv',
]
data_path = next((path.resolve() for path in candidate_paths if path.exists()), None)
if data_path is None:
    raise FileNotFoundError('Place employees.csv in project/data/')

employees = pd.read_csv(data_path)
print(f'Loaded: {data_path.name}')
employees.shape
```
